# 01 - Kiểm thử Module Audio
Kiểm tra việc đọc file âm thanh (.wav), nạp nhãn Ground Truth (.lab) và tạo Dataset.


```python
import sys
from pathlib import Path

# Thêm đường dẫn gốc dự án
ROOT_DIR = Path.cwd().resolve()
while ROOT_DIR.name != 'main_gk' and ROOT_DIR.parent != ROOT_DIR:
    ROOT_DIR = ROOT_DIR.parent
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.audio.loader import load_audio
from src.audio.labels import parse_lab, create_frame_labels
from src.audio.dataset import load_dataset
```

### 1. Kiểm tra đọc một file âm thanh đơn lẻ


```python
wav_file = ROOT_DIR / 'TinHieuKiemThu' / 'studio_M2.wav'
load = load_audio(wav_file)

print(f'Tên file      : {wav_file.name}')
print(f'Sampling rate : {load.fs} Hz')
print(f'Thời lượng    : {load.duration_sec:.2f} giây')
print(f'Miền biên độ  : [{load.signal.min():.3f}, {load.signal.max():.3f}]')
```

### 2. Kiểm tra đọc file nhãn Ground Truth (.lab)


```python
lab_file = ROOT_DIR / 'TinHieuKiemThu' / 'studio_M2.lab'
lab_data = parse_lab(lab_file)

print(f'Mốc biên chuẩn (boundaries)  : {lab_data["boundaries"]}')
print(f'Đoạn tiếng nói (speech_segs) : {lab_data["speech_segments"]}')
print(f'F0 mean / std                 : {lab_data["f0_mean"]} / {lab_data["f0_std"]}')
```

### 3. Kiểm tra nạp toàn bộ tập dữ liệu (Dataset)


```python
signals = load_dataset(ROOT_DIR / 'TinHieuKiemThu')
print(f'Đã nạp thành công {len(signals)} file:')
for s in signals:
    env = 'Phone' if s.is_phone else 'Studio'
    print(f' - {s.name:<12} | {env:<7} | Duration: {s.duration_sec:.2f}s | Biên GT: {s.gt_boundaries}')
```
